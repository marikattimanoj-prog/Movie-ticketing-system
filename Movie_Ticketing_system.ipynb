```python
import os

class MovieManager:
    movie_file = "movies.txt"

    def add_movie(self):
        name = input("Enter Movie Name: ")
        language = input("Enter Language: ")
        genre = input("Enter Genre: ")

        with open(self.movie_file, "a") as f:
            f.write(name + "," + language + "," + genre + "\n")

        print("Movie added successfully!")

    def view_movies(self):
        if not os.path.exists(self.movie_file):
            print("No movies found.")
            return

        print("\nMovie List:")
        with open(self.movie_file, "r") as f:
            for line in f:
                data = line.strip().split(",")

                # SAFEST POSSIBLE CHECK
                if len(data) >= 3:
                    print("Name:", data[0])
                    print("Language:", data[1])
                    print("Genre:", data[2])
                    print("-" * 20)


class TicketManager:
    ticket_file = "tickets.txt"

    def add_ticket(self):
        booking_id = input("Enter Booking ID: ")
        customer = input("Enter Customer Name: ")
        movie = input("Enter Movie Name: ")
        date = input("Enter Date: ")
        time = input("Enter Show Time: ")
        ttype = input("Enter Ticket Type: ")
        seat = input("Enter Seat Number: ")

        with open(self.ticket_file, "a") as f:
            f.write(
                booking_id + "," + customer + "," + movie + "," +
                date + "," + time + "," + ttype + "," + seat + "\n"
            )

        print("Ticket booked successfully!")

    def view_tickets(self):
        if not os.path.exists(self.ticket_file):
            print("No tickets found.")
            return

        print("\nTicket Records:")
        with open(self.ticket_file, "r") as f:
            for line in f:
                data = line.strip().split(",")
                if len(data) >= 7:
                    print(data)


def main():
    movie_manager = MovieManager()
    ticket_manager = TicketManager()

    while True:
        print("\n--- Movie Ticketing System ---")
        print("1. Add Movie")
        print("2. View Movies")
        print("3. Book Ticket")
        print("4. View Tickets")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            movie_manager.add_movie()
        elif choice == "2":
            movie_manager.view_movies()
        elif choice == "3":
            ticket_manager.add_ticket()
        elif choice == "4":
            ticket_manager.view_tickets()
        elif choice == "5":
            print("Thank you!")
            break
        else:
            print("Invalid choice")


main()
```
